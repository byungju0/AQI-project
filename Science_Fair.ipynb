```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import Lasso
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import AdaBoostRegressor
from xgboost.sklearn import XGBRegressor

#Reads the combined Beijing PM2.5 data
data = pd.read_csv("/content/drive/My Drive/data/Science Fair Data Collection/folder/BeijingPMTest.csv",names = ["PM2.5","date"])

#Remove outliers
data=data[data['PM2.5'] <= 250]

#Deletes and fills lines with N/A values
data.fillna(method ='ffill', inplace = True)
data.dropna(inplace = True)

#sets X and y values as date and PM2.5 values
y = data['PM2.5']
X = data.drop(['PM2.5'], axis=1)
X_value=data['date']
Y_value=data['PM2.5']

#Splits the training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3,random_state=1200)

#Creates a linear regression model that will be trained
lin_regression = LinearRegression()

#Trains the model using the train values
lin_regression.fit(X_train,y_train)

#Predicts values with trained data
lin_prediction = lin_regression.predict(X_test)

#Gives the score of the trained regression by comparing with actual value
lin_regression_score=lin_regression.score(X_test,y_test)
print(lin_regression_score)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.scatter(y_test,lin_prediction)
plt.title('Linear Regression')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.show()

#plotting the graph of the linear regression
X_test_copied = X_test.copy()
X_test_copied = range(len(y_test))

plt.figure(figsize=(30,10))
plt.title('Linear Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,lin_prediction,color='r')
plt.show()

#Creates a lasso regression model that will be trained
lasso_regression = Lasso()

#Trains the model using the train values
lasso_regression.fit(X_train,y_train)

#Gives the score of the trained regression by comparing with actual value
lasso_score=lasso_regression.score(X_test,y_test)
print(lasso_score)

#Predicts values with trained data
lasso_prediction=lasso_regression.predict(X_test)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.title('Lasso Regression')
plt.scatter(y_test,lasso_prediction)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the lasso regression
plt.figure(figsize=(30,10))
plt.title('Lasso Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,lasso_prediction,color='r')
plt.show()

#Creates a Random Forest regression model that will be trained
randomforest_regression = RandomForestRegressor(n_estimators=350,random_state=1200)

#Trains the model using the train values
randomforest_regression.fit(X_train,y_train)

#Gives the score of the trained regression by comparing with actual value
randomforest_score=randomforest_regression.score(X_test,y_test)
print(randomforest_score)

#Predicts values with only trained data
randomforest_prediction = randomforest_regression.predict(X_test)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.title('Random Forest Regression')
plt.scatter(y_test,randomforest_prediction)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the randomforest regression
plt.figure(figsize=(30,10))
plt.title('Random Forest Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,randomforest_prediction,color='r')
plt.show()

#Creates a Extra Trees regression model that will be trained
extra_tree_reg=ExtraTreesRegressor(n_estimators=400,random_state=800)

#Trains the model using the train values
extra_tree_reg.fit(X_train,y_train)

#Gives the score of the trained regression by comparing with actual value
extra_tree_score=extra_tree_reg.score(X_test,y_test)
print(extra_tree_score)

#Predicts values with only trained data
extratree_prediction= extra_tree_reg.predict(X_test)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.title('Extra Tree Regression')
plt.scatter(y_test,extratree_prediction)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the extratree regression
X_test_copied = X_test.copy()
X_test_copied = range(len(y_test))

plt.figure(figsize=(30,10))
plt.title('Extra Tree Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,extratree_prediction,color='r')
plt.show()

#Creates a Decision Tree regression model that will be trained
tree = DecisionTreeRegressor()

#Creates a AdaBoost regression model that will be alter the decision tree regression model
boosted_tree = AdaBoostRegressor(DecisionTreeRegressor(),n_estimators=300,learning_rate=.1)

#Trains both models using the train values
tree.fit(X_train,y_train)
boosted_tree.fit(X_train,y_train)

#Gives the score of the trained regression by comparing with actual value
print(tree.score(X_test,y_test))
boosted_tree_score=boosted_tree.score(X_test,y_test)
print(boosted_tree_score)

#Predicts values with trained data
tree_prediction=tree.predict(X_test)
boosted_tree_prediction=boosted_tree.predict(X_test)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.title('Decision Tree Regression')
plt.scatter(y_test,tree_prediction)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the decisiontree regression
plt.figure(figsize=(30,10))
plt.title('Decision Tree Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,tree_prediction,color='r')
plt.show()

plt.figure(figsize=(10,10))
plt.title('Boosted Decision Tree Regression')
plt.scatter(y_test,boosted_tree_prediction)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the boosted decision regression
plt.figure(figsize=(30,10))
plt.title('Boosted Decision Tree Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,boosted_tree_prediction,color='r')
plt.show()

#Creates a XGB regression model that will be trained
xboost = XGBRegressor(n_estimators=250)

#Trains the model using the train values
xboost.fit(X_train,y_train)

#Gives the score of the trained regression by comparing with actual value
xgb_score=xboost.score(X_test,y_test)
print(xgb_score)

#Predicts values with trained data
xboost_pred = xboost.predict(X_test)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.title('XGB Regression')
plt.scatter(y_test,xboost_pred)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the XGB regression
plt.figure(figsize=(30,10))
plt.title('XGB Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,boosted_tree_prediction,color='r')
plt.show()



```

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPRegressor
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import Lasso
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import AdaBoostRegressor
from xgboost.sklearn import XGBRegressor

#Reads the combined Tianjin PM2.5 data
data = pd.read_csv("/content/drive/My Drive/data/Science Fair Data Collection/folder/TianjinDates.csv",names = ["PM2.5","date"])

#Remove outliers
data=data[data['PM2.5'] <= 250]

#Deletes and fills lines with N/A values
data.fillna(method ='ffill', inplace = True)
data.dropna(inplace = True)

#sets X and y values as date and PM2.5 values
y = data['PM2.5']
X = data.drop(['PM2.5'], axis=1)
X_value=data['date']
Y_value=data['PM2.5']

#Splits the training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3,random_state=1200)

#Creates a linear regression model that will be trained
lin_regression = LinearRegression()

#Trains the model using the train values
lin_regression.fit(X_train,y_train)

#Predicts values with trained data
lin_prediction = lin_regression.predict(X_test)

#Gives the score of the trained regression by comparing with actual value
lin_regression_score=lin_regression.score(X_test,y_test)
print(lin_regression_score)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.scatter(y_test,lin_prediction)
plt.title('Linear Regression')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.show()

#plotting the graph of the linear regression
X_test_copied = X_test.copy()
X_test_copied = range(len(y_test))

plt.figure(figsize=(30,10))
plt.title('Linear Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,lin_prediction,color='r')
plt.show()

#Creates a MLP regression model that will be trained
mlp_regression = MLPRegressor()

#Trains the model using the train values
mlp_regression.fit(X_train,y_train)

#Gives the score of the trained regression by comparing with actual value
score = mlp_regression.score(X_test,y_test)
print(score)

#Predicts values with trained data
mlp_prediction=mlp_regression.predict(X_test)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.title('MLP Regression')
plt.scatter(y_test,mlp_prediction)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the mlp regression
plt.figure(figsize=(30,10))
plt.title('MLP Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,mlp_prediction,color='r')
plt.show()

#Creates a lasso regression model that will be trained
lasso_regression = Lasso()

#Trains the model using the train values
lasso_regression.fit(X_train,y_train)

#Gives the score of the trained regression by comparing with actual value
lasso_score=lasso_regression.score(X_test,y_test)
print(lasso_score)

#Predicts values with trained data
lasso_prediction=lasso_regression.predict(X_test)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.title('Lasso Regression')
plt.scatter(y_test,lasso_prediction)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the lasso regression
plt.figure(figsize=(30,10))
plt.title('Lasso Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,lasso_prediction,color='r')
plt.show()

#Creates a Random Forest regression model that will be trained
randomforest_regression = RandomForestRegressor(n_estimators=350,random_state=1200)

#Trains the model using the train values
randomforest_regression.fit(X_train,y_train)

#Gives the score of the trained regression by comparing with actual value
randomforest_score=randomforest_regression.score(X_test,y_test)
print(randomforest_score)

#Predicts values with only trained data
randomforest_prediction = randomforest_regression.predict(X_test)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.title('Random Forest Regression')
plt.scatter(y_test,randomforest_prediction)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the randomforest regression
plt.figure(figsize=(30,10))
plt.title('Random Forest Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,randomforest_prediction,color='r')
plt.show()

#Creates a Extra Trees regression model that will be trained
extra_tree_reg=ExtraTreesRegressor(n_estimators=400,random_state=800)

#Trains the model using the train values
extra_tree_reg.fit(X_train,y_train)

#Gives the score of the trained regression by comparing with actual value
extra_tree_score=extra_tree_reg.score(X_test,y_test)
print(extra_tree_score)

#Predicts values with only trained data
extratree_prediction= extra_tree_reg.predict(X_test)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.title('Extra Tree Regression')
plt.scatter(y_test,extratree_prediction)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the extratree regression
X_test_copied = X_test.copy()
X_test_copied = range(len(y_test))

plt.figure(figsize=(30,10))
plt.title('Extra Tree Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,extratree_prediction,color='r')
plt.show()

#Creates a Decision Tree regression model that will be trained
tree = DecisionTreeRegressor()
#Creates a AdaBoost regression model that will be alter the decision tree regression model
boosted_tree = AdaBoostRegressor(DecisionTreeRegressor(),n_estimators=300,learning_rate=.1)

#Trains both models using the train values
tree.fit(X_train,y_train)
boosted_tree.fit(X_train,y_train)

#Gives the score of the trained regression by comparing with actual value
print(tree.score(X_test,y_test))
boosted_tree_score=boosted_tree.score(X_test,y_test)
print(boosted_tree_score)

#Predicts values with trained data
tree_prediction=tree.predict(X_test)
boosted_tree_prediction=boosted_tree.predict(X_test)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.title('Decision Tree Regression')
plt.scatter(y_test,tree_prediction)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the decisiontree regression
plt.figure(figsize=(30,10))
plt.title('Decision Tree Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,tree_prediction,color='r')
plt.show()

plt.figure(figsize=(10,10))
plt.title('Boosted Decision Tree Regression')
plt.scatter(y_test,boosted_tree_prediction)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()

#plotting the graph of the boosted decision regression
plt.figure(figsize=(30,10))
plt.title('Boosted Decision Tree Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,boosted_tree_prediction,color='r')
plt.show()

#Creates a XGB regression model that will be trained
xboost = XGBRegressor(n_estimators=250)

#Trains the model using the train values
xboost.fit(X_train,y_train)

#Gives the score of the trained regression by comparing with actual value
xgb_score=xboost.score(X_test,y_test)
print(xgb_score)

#Predicts values with trained data
xboost_pred = xboost.predict(X_test)

#Plots the graph of actual data vs predicted data
plt.figure(figsize=(10,10))
plt.title('XGB Regression')
plt.scatter(y_test,xboost_pred)
plt.xlabel('Actual')
plt.ylabel('predicted')
plt.show()


#plotting the graph of the XGB regression
plt.figure(figsize=(30,10))
plt.title('XGB Regression')
plt.scatter(X_test_copied,y_test,color='k')
plt.plot(X_test_copied,y_test,color='orange',linestyle='--')
plt.plot(X_test_copied,boosted_tree_prediction,color='r')
plt.show()
```
